## Exercício 13

**(c)** Gere (pelo Python) 200 números pseudo-aleatórios no intervalo (0, 1). Utilize esses números e a função de distribuição F_Y para gerar 200 observações de Y ~ Exponencial(0, 16).

```python
import numpy as np


def uniform01(n):
    return np.random.uniform(0, 1, n)


def exponential(U):
    return -np.log(1 - U) / 0.16


U = uniform01(200)
Y = exponential(U)

Y
```

**(d)** Faça um histograma dessas 200 observações. Fazer em Python ou R. Compare com a f.d.p. f_Y (pode fazer gráfico da f.d.p. sobre o histograma). Comente.

```python
import matplotlib.pyplot as plt

plt.hist(Y, bins=15, density=True, alpha=0.6, edgecolor="black", label="Observações")

y = np.linspace(0, max(Y), 500)
fY = 0.16 * np.exp(-0.16 * y)

plt.plot(y, fY, linewidth=2, label=r"$f_Y(y)=0.16e^{-0.16y}$")

plt.xlabel("y")
plt.ylabel("Densidade")
plt.title("Histograma das 200 observações de Y")
plt.legend()
```
